# GUI auxiliar de agendamento

```python
import tkinter as tk
from tkcalendar import Calendar

def selecionar_data():
    data = cal.get_date()
    print("Compromisso em:", data)


root = tk.Tk()
root.title("Selecionar Data")


cal = Calendar(root, selectmode="day")
cal.pack(pady=20)

btn = tk.Button(root, text="Salvar Compromisso", command=selecionar_data)
btn.pack()


root.mainloop()
```
