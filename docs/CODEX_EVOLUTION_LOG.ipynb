# Log de Evolucoes - ScoreIA (Codex)

Registro cronologico de alteracoes, testes e validacoes.


## Template Padrao

- `timestamp`: data/hora da alteracao
- `change`: resumo objetivo da mudanca
- `files`: arquivos impactados
- `validation`: validacao executada

Exemplo:
- timestamp: `2026-03-04 10:00:00`
- change: ajuste de threshold otimizado em classificacao
- files: `tools/metrics.py`, `tools/optimization.py`
- validation: parse sintatico + smoke test
