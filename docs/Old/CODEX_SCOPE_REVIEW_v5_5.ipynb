# ScoreIA — Leitura de Escopo (Fase 5.5)

## Entendimento consolidado

1. O projeto está formalmente na **Fase 5.5**, com pipeline de ML já consolidado em termos de separação treino/teste, CV manual, otimização com Optuna e modo `auto` para competição entre LightGBM e MLP.
2. A arquitetura vigente é **modular por responsabilidade** e deve ser preservada:
   - `core/agent.py`: orquestração
   - `tools/data_pipeline.py`: dados e transformações
   - `tools/optimization.py`: treino/CV/otimização
   - `tools/model_factory.py`: construção estrutural dos modelos
   - `tools/metrics.py`: métricas e padronização de resultados
3. O documento de identidade reforça que ScoreIA é um sistema evolutivo por fases, com postura de controle, privacidade e rastreabilidade (não-autonomia irrestrita).
4. O objetivo imediato da fase atual é fechar três frentes:
   - **Interpretabilidade** (SHAP + feature importance)
   - **Exportação** (artefatos de modelo/pipeline)
   - **Relatório técnico automático** (saída consolidada para comunicação de resultados)
5. Há uma pendência estrutural já reconhecida para `main.py`: avançar para execução mais unificada (CLI e evolução para backend/web), sem quebrar modularidade.

## Parecer técnico sobre a estrutura

- A arquitetura está bem orientada para escalabilidade incremental e reduz risco de acoplamento indevido.
- O pipeline segue boas práticas estatísticas explícitas (holdout antes de CV e teste intocável).
- A separação entre camada de decisão (agent) e camada de modelagem (tools) está alinhada com manutenção de longo prazo.
- O estado atual é adequado para implementar interpretabilidade/exportação/relatório sem refatoração arquitetural ampla.
