# ScoreIA — Codex Context Rules

Versão: 5.5
Modo: Continuidade Privada Controlada

---

# 1. Objetivo

Este documento define regras obrigatórias que o Codex deve respeitar ao sugerir alterações no projeto ScoreIA.

Ele existe para:

* Evitar deriva arquitetural
* Preservar consistência estatística
* Manter modularidade
* Garantir evolução controlada
* Impedir refatorações estruturais indevidas

Este documento é normativo.

---

# 2. Princípios Estruturais Imutáveis

## 2.1 Separação de Responsabilidades

É proibido misturar responsabilidades entre módulos.

### Estrutura obrigatória:

* `agent.py` → Orquestração
* `data_pipeline.py` → Engenharia de dados
* `optimization.py` → Treino e cross-validation
* `model_factory.py` → Construção estrutural de modelos
* `metrics.py` → Métricas
* `webapp/` → Interface
* `main.py` → Entrada CLI

Nenhuma lógica estatística deve migrar para `agent.py`.

---

## 2.2 Regra de Ouro: Teste é Intocável

O conjunto de teste:

* Nunca participa da otimização
* Nunca participa da validação cruzada
* Nunca participa de early stopping
* Nunca é usado para ajuste de hiperparâmetros

Qualquer sugestão que viole isso deve ser rejeitada.

---

## 2.3 Ordem Obrigatória do Pipeline

O fluxo oficial é:

```text
Dataset original
    ↓
Holdout (train/test)
    ↓
Cross-validation (apenas treino)
    ↓
Otimização
    ↓
Treino final
    ↓
Avaliação no teste
```

Não alterar essa ordem.

---

## 2.4 Webapp Não Contém Lógica de ML

O diretório `tools/webapp/`:

* Apenas dispara execução
* Apenas coleta parâmetros
* Apenas exibe resultados

Não pode conter:

* Treino
* Cross-validation
* Otimização
* Engenharia de dados

---

## 2.5 Agent Não Modela

`core/agent.py`:

* Não deve conter loops de treino
* Não deve conter lógica de CV
* Não deve conter otimização
* Não deve conter engenharia de dados

Ele apenas orquestra módulos.

---

# 3. Regras para Sugestões do Codex

Ao propor qualquer alteração, o Codex deve:

1. Explicar impacto arquitetural.
2. Confirmar que não viola o ML_PIPELINE_SPEC.md.
3. Confirmar que não cria data leakage.
4. Confirmar que mantém modularidade.
5. Não propor simplificações que misturem responsabilidades.

---

# 4. Restrições de Refatoração

Refatorações são permitidas apenas se:

* Melhorarem legibilidade
* Reduzirem duplicação
* Não alterarem ordem do pipeline
* Não alterarem isolamento do teste
* Não alterarem responsabilidades

Refatorações estruturais profundas exigem novo snapshot arquitetural.

---

# 5. Regras para Novos Recursos

Novos recursos devem:

* Ser implementados como módulos independentes
* Não modificar funções já consolidadas
* Não alterar contrato de retorno do pipeline
* Não modificar fluxo oficial de dados

Exemplos corretos:

* `interpretability.py`
* `model_export.py`
* `report_generator.py`

Exemplos incorretos:

* Inserir SHAP dentro de optimization.py
* Inserir exportação dentro de agent.py
* Misturar geração de PDF com treino

---

# 6. Política para Fase Atual (5.5)

Prioridade atual:

1. SHAP + Feature Importance
2. Exportação de modelo
3. Relatório técnico automático
4. Revisão estrutural do main.py

O Codex não deve:

* Introduzir novos modelos
* Alterar estratégia de validação
* Trocar framework de otimização
* Alterar holdout

---

# 7. Política de Evolução

ScoreIA está em:

Modo: Continuidade Privada Controlada.

Isso significa:

* Estabilidade > inovação radical
* Coerência > experimentação
* Consolidação > expansão

Mudanças estruturais só após finalização completa da Fase 5.5.

---

# 8. Compatibilidade Futura

O projeto deve permanecer compatível com:

* Migração para backend FastAPI
* Evolução para SaaS
* Possível abertura open-source
* Modularização em pacote instalável

Evitar acoplamentos desnecessários.

---

# 9. Proibições Explícitas

O Codex não deve sugerir:

* Usar teste para early stopping global
* Treinar modelo final antes da otimização
* Fundir data_pipeline com optimization
* Colocar lógica estatística na webapp
* Remover cross-validation
* Simplificar pipeline para “fit direto”
